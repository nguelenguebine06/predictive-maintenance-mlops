#  Contrôle de la qualité des données

## Objectif

L'objectif de cette étape est d'évaluer la qualité structurelle du dataset MetroPT-3 avant de poursuivre l'analyse exploratoire.

Nous cherchons à identifier les éventuels problèmes susceptibles d'affecter les analyses statistiques et les futurs modèles de Machine Learning.

##  Contrôles réalisés

Les contrôles portent sur :

- Les valeurs manquantes : identifier les données absentes.
- Les doublons : détecter les observations répétées.
- Les types de données : vérifier la nature des variables.
- Les variables constantes : identifier les colonnes sans variation.
- Les valeurs infinies : détecter les valeurs numériques non finies.

##  Méthodologie

Les contrôles seront réalisés avec Pandas et NumPy.

Aucune suppression ou modification des données ne sera effectuée durant cette étape.


```python
# 1. Valeurs manquantes
missing_values = df.isnull().sum()

# 2. Lignes entièrement dupliquées
duplicate_rows = df.duplicated().sum()

# 3. Types de données
data_types = df.dtypes.astype(str)

# 4. Variables constantes
constant_columns = [
    col for col in df.columns
    if df[col].nunique(dropna=False) <= 1
]

# 5. Valeurs infinies dans les colonnes numériques
numeric_df = df.select_dtypes(include=np.number)

infinite_values = np.isinf(numeric_df).sum()

# 6. Construction du tableau de synthèse
quality_report = pd.DataFrame({
    "Type de contrôle": [
        "Valeurs manquantes",
        "Lignes dupliquées",
        "Valeurs infinies",
        "Variables constantes"
    ],
    "Résultat": [
        int(missing_values.sum()),
        int(duplicate_rows),
        int(infinite_values.sum()),
        len(constant_columns)
    ]
})

# 7. Affichage des résultats
print("RAPPORT DE QUALITÉ DES DONNÉES")
display(quality_report)

print("\nVALEURS MANQUANTES PAR VARIABLE")
display(missing_values.to_frame(name="Nombre de valeurs manquantes"))

print("\nTYPES DES VARIABLES")
display(data_types.to_frame(name="Type"))

print("\nVARIABLES CONSTANTES")
print(constant_columns)

print("\nVALEURS INFINIES PAR VARIABLE")
display(infinite_values.to_frame(name="Nombre de valeurs infinies"))
```


###  Conclusion du contrôle de qualité

Les contrôles réalisés sur le dataset MetroPT-3 montrent les résultats suivants :

- Aucune valeur manquante détectée.
- Aucun doublon complet détecté.
- Aucune valeur infinie détectée.
- Aucune variable constante identifiée.

Concernant les types de données, le dataset contient 15 variables numériques de type float64, une variable entière de type int64 et une variable textuelle de type str correspondant au timestamp.

Le dataset présente donc une bonne qualité structurelle selon les contrôles effectués.

Cependant, ces résultats ne garantissent pas l'absence d'erreurs de mesure ou de valeurs physiquement incohérentes. Des analyses complémentaires seront nécessaires.



#  Analyse temporelle

##  Objectif

L'objectif de cette analyse est d'étudier la dimension temporelle du dataset MetroPT-3.

Nous cherchons à comprendre la période couverte par les observations, la fréquence d'acquisition des mesures et la présence éventuelle d'interruptions dans l'enregistrement des données.

## Méthodologie

L'analyse repose sur les contrôles suivants :

- Conversion de la variable timestamp en format datetime.
- Identification des dates de début et de fin.
- Vérification de l'ordre chronologique.
- Calcul des intervalles entre deux observations successives.
- Analyse de la fréquence d'acquisition.
- Identification des interruptions temporelles.

Aucune observation ne sera supprimée ou modifiée durant cette analyse.


```python
# 1. Conversion des horodatages
timestamps = pd.to_datetime(df["timestamp"], errors="coerce")

# 2. Informations générales sur la période
date_debut = timestamps.min()
date_fin = timestamps.max()
duree = date_fin - date_debut

print("PERIODE COUVERTE")
print("Date de début :", date_debut)
print("Date de fin :", date_fin)
print("Durée totale :", duree)

# 3. Vérification de la chronologie
print("\nCONTROLE CHRONOLOGIQUE")
print("Dates invalides :", timestamps.isna().sum())
print("Ordre chronologique :", timestamps.is_monotonic_increasing)
print("Horodatages dupliqués :", timestamps.duplicated().sum())

# 4. Calcul des intervalles entre observations
intervalles = timestamps.diff().dt.total_seconds().dropna()

print("\nSTATISTIQUES DES INTERVALLES (secondes)")
display(intervalles.describe().to_frame("Valeur"))

# 5. Fréquence des intervalles les plus courants
print("\nINTERVALLES LES PLUS FREQUENTS")
display(
    intervalles.value_counts()
    .head(10)
    .rename_axis("Intervalle (secondes)")
    .to_frame("Nombre d'observations")
)

# 6. Classification des interruptions
seuils = {
    "Supérieures à 10 secondes": 10,
    "Supérieures à 1 minute": 60,
    "Supérieures à 10 minutes": 600,
    "Supérieures à 1 heure": 3600,
    "Supérieures à 24 heures": 86400
}

interruptions_report = pd.DataFrame({
    "Catégorie": seuils.keys(),
    "Nombre": [
        int((intervalles > seuil).sum())
        for seuil in seuils.values()
    ]
})

print("\nCLASSIFICATION DES INTERRUPTIONS")
display(interruptions_report)

# 7. Identification des 10 plus grandes interruptions
interruptions = pd.DataFrame({
    "Date précédente": timestamps.shift(1),
    "Date actuelle": timestamps,
    "Intervalle (secondes)": intervalles.reindex(timestamps.index)
})

top_interruptions = (
    interruptions[interruptions["Intervalle (secondes)"] > 60]
    .nlargest(10, "Intervalle (secondes)")
    .copy()
)

top_interruptions["Durée (heures)"] = (
    top_interruptions["Intervalle (secondes)"] / 3600
)

print("\nLES 10 PLUS GRANDES INTERRUPTIONS")
display(top_interruptions)
```

```python
# Distribution des intervalles inférieurs ou égaux à 60 secondes
intervalles_courts = intervalles[intervalles <= 60]

plt.figure(figsize=(10, 5))

sns.histplot(
    intervalles_courts,
    bins=30,
    kde=False
)

plt.title("Distribution des intervalles entre les mesures")
plt.xlabel("Intervalle entre deux mesures (secondes)")
plt.ylabel("Nombre d'observations")

plt.tight_layout()
plt.savefig(f"Distribution_des_intervales_d'observations.png")
plt.show()
```


###  Conclusion de l'analyse temporelle

L'analyse temporelle montre que le dataset MetroPT-3 couvre la période du 1er février 2020 au 1er septembre 2020.

La fréquence médiane d'acquisition est de 10 secondes. Toutefois, les intervalles entre les observations ne sont pas parfaitement réguliers.

331 intervalles dépassent une minute, dont 160 dépassent une heure et 5 dépassent 24 heures. La plus longue interruption observée est d'environ 48 heures.

Ces résultats mettent en évidence la présence d'interruptions dans l'acquisition des données.

Ces interruptions devront être prises en compte lors des futures étapes de préparation des données et de modélisation temporelle. Aucune observation n'a été supprimée durant cette analyse.

#  Analyse bivariée et multivariée

##  Objectif

L'objectif de cette étape est d'étudier les relations statistiques entre les différentes variables du dataset MetroPT-3.

Contrairement à l'analyse univariée, qui étudie chaque variable séparément, l'analyse bivariée et multivariée permet d'observer les relations entre plusieurs variables.

##  Méthodologie

L'analyse sera organisée en plusieurs étapes :

- Étude des corrélations entre les variables numériques.
- Visualisation des corrélations avec une matrice thermique (heatmap).
- Analyse graphique des relations entre certains capteurs.
- Étude des relations entre les variables de mesure et les variables binaires.



###  Matrice de corrélation de Pearson

Le coefficient de corrélation de Pearson permet de mesurer l'intensité et le sens de la relation linéaire entre deux variables numériques.

Une valeur positive indique que les variables ont tendance à évoluer dans le même sens.

Une valeur négative indique qu'elles ont tendance à évoluer en sens opposé.

Une valeur proche de zéro indique une faible relation linéaire, sans exclure l'existence d'une relation non linéaire.


```python
# Corrélation des variables analogiques de mesure
sensor_columns = [
    "TP2",
    "TP3",
    "H1",
    "DV_pressure",
    "Reservoirs",
    "Oil_temperature",
    "Motor_current"
]

# Calcul de la matrice de corrélation
correlation_matrix = df[sensor_columns].corr(method="pearson")

# Affichage de la matrice
display(correlation_matrix)
```


### Visualisation des corrélations

La matrice de corrélation sera représentée sous forme de heatmap.

Cette visualisation permet d'identifier rapidement les associations linéaires positives et négatives entre les variables étudiées.


```python
# Visualisation de la matrice de corrélation

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True
)

plt.title("Matrice de corrélation des capteurs")
plt.tight_layout()
plt.savefig(f"Matrice_de_corrélation_variables_analogiques.png")
plt.show()
```


###  Identification des corrélations les plus fortes

Cette étape permet de classer les paires de variables selon l'intensité de leur corrélation linéaire.

L'objectif est de sélectionner les relations qui méritent une analyse graphique plus approfondie.


```python
# Extraction des corrélations sans doublons
correlation_pairs = (
    correlation_matrix
    .where(
        np.triu(
            np.ones(correlation_matrix.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .reset_index()
)

correlation_pairs.columns = [
    "Variable 1",
    "Variable 2",
    "Corrélation"
]

# Classement par intensité absolue
correlation_pairs["Intensité"] = (
    correlation_pairs["Corrélation"].abs()
)

correlation_pairs = correlation_pairs.sort_values(
    by="Intensité",
    ascending=False
)

display(correlation_pairs.head(10).round(3))
```


###  Vérification de la redondance entre TP3 et Reservoirs

La matrice de corrélation révèle une corrélation de 1,000 entre TP3 et Reservoirs.

Une vérification complémentaire est nécessaire pour déterminer si ces variables contiennent des valeurs identiques ou simplement des valeurs fortement corrélées.

Cette vérification permettra d'identifier une éventuelle redondance entre les variables.


```python
# Vérification de l'identité des deux variables

print(
    "Colonnes strictement identiques :",
    df["TP3"].equals(df["Reservoirs"])
)

# Différence entre les deux variables
difference = df["TP3"] - df["Reservoirs"]

print("\nSTATISTIQUES DES DIFFERENCES")
display(difference.describe().to_frame("TP3 - Reservoirs"))

print(
    "\nNombre de différences non nulles :",
    (difference != 0).sum()
)

# Corrélation avec davantage de décimales
print(
    "\nCorrélation précise :",
    df["TP3"].corr(df["Reservoirs"])
)
```


###  Analyse graphique des relations entre capteurs

Les nuages de points permettent de visualiser les relations entre deux variables numériques.

Un échantillon de 10 000 observations sera utilisé pour rendre les graphiques plus lisibles.

Les relations étudiées sont :

- TP2 et H1.
- TP2 et Motor_current.
- Oil_temperature et Motor_current.
- TP3 et Reservoirs.

L'objectif est de vérifier visuellement les tendances observées dans la matrice de corrélation.


```python
# Échantillonnage pour la visualisation
df_sample = df.sample(
    n=10000,
    random_state=42
)

# Paires de variables à étudier
pairs = [
    ("TP2", "H1"),
    ("TP2", "Motor_current"),
    ("Oil_temperature", "Motor_current"),
    ("TP3", "Reservoirs")
]

# Création des graphiques
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes = axes.flatten()

for i, (x, y) in enumerate(pairs):

    sns.scatterplot(
        data=df_sample,
        x=x,
        y=y,
        alpha=0.25,
        s=12,
        ax=axes[i]
    )

    axes[i].set_title(f"Relation entre {x} et {y}")

plt.tight_layout()
plt.savefig(f"Scatterplot_variables_correlées.png")
plt.show()
```


###  Conclusion de l'analyse bivariée et multivariée

L'analyse des corrélations et des nuages de points a permis d'identifier plusieurs relations entre les variables du dataset MetroPT-3.

Les principales observations sont :

- TP3 et Reservoirs présentent une corrélation extrêmement forte de 0,9999934. Les deux colonnes ne sont toutefois pas strictement identiques.
- TP2 et H1 présentent une forte corrélation négative de -0,961.
- TP2 et Motor_current présentent une corrélation positive de 0,697.
- Oil_temperature et Motor_current présentent une corrélation positive de 0,529.
- Plusieurs nuages de points révèlent des concentrations distinctes pouvant correspondre à différents régimes de fonctionnement.

Ces résultats montrent que certaines variables évoluent de manière étroitement liée.

Cependant, les corrélations globales ne suffisent pas à expliquer les relations physiques entre les capteurs. Les différents états de fonctionnement du compresseur devront être pris en compte dans les analyses complémentaires.



###  Analyse des capteurs selon les variables binaires

#### Objectif

L'objectif est d'étudier les variations des mesures des capteurs selon les valeurs des variables binaires du dataset.

Cette analyse permettra de rechercher d'éventuelles différences de comportement entre les observations associées aux valeurs 0 et 1.

#### Variables étudiées

Les variables binaires retenues sont :

- COMP
- DV_eletric
- Towers
- MPG
- LPS
- Pressure_switch
- Oil_level
- Caudal_impulses

#### Méthodologie

L'analyse sera réalisée en deux étapes :

1. Comparaison des statistiques descriptives des capteurs selon les valeurs binaires.
2. Visualisation de certaines relations à l'aide de boxplots.

Les valeurs 0 et 1 seront considérées comme des modalités numériques, sans leur attribuer de signification physique non vérifiée.



#### Comparaison statistique des capteurs

Pour chaque variable binaire, les observations sont séparées selon les modalités 0 et 1.

Les statistiques descriptives permettent de comparer les distributions des mesures entre ces deux groupes.

Cette comparaison peut révéler des différences de comportement associées aux modalités des variables binaires.


```python
for binary_col in binary_columns:

    print(f"\n{'=' * 60}")
    print(f"ANALYSE DE LA VARIABLE : {binary_col}")
    print(f"{'=' * 60}")

    grouped_stats = (
        df.groupby(binary_col)[sensor_columns]
        .agg(["count", "mean", "median", "std"])
        .round(3)
    )

    display(grouped_stats)
```


#### Analyse graphique selon les variables binaires

Les boxplots permettent de comparer la distribution des mesures des capteurs entre les modalités 0 et 1.

Un échantillon aléatoire sera utilisé pour faciliter la représentation graphique.

Les graphiques permettront d'observer les différences de médiane, de dispersion et la présence éventuelle de valeurs atypiques entre les groupes.


```python
# Variables binaires
selected_binary = [
    "COMP",
    "DV_eletric",
    "LPS"
]

# Échantillon pour les visualisations
df_sample_binary = df.sample(
    n=20000,
    random_state=42
)

# Création des graphiques
fig, axes = plt.subplots(
    len(selected_binary),
    len(sensor_columns),
    figsize=(21, 12)
)

for i, binary_col in enumerate(selected_binary):

    for j, sensor_col in enumerate(sensor_columns):

        sns.boxplot(
            data=df_sample_binary,
            x=binary_col,
            y=sensor_col,
            ax=axes[i, j],
            showfliers=False
        )

        axes[i, j].set_title(
            f"{sensor_col} selon {binary_col}"
        )

        axes[i, j].set_xlabel(binary_col)
        axes[i, j].set_ylabel(sensor_col)

plt.tight_layout()
plt.show()
```


###  Analyse des associations entre variables binaires

Les premières comparaisons révèlent des comportements similaires entre COMP et DV_eletric.

Cette étape vise à mesurer leur association statistique et à étudier la répartition de LPS.

Les effectifs complets seront utilisés afin de ne pas introduire de biais lié à l'échantillonnage.


```python
# 1. Tableau croisé COMP / DV_eletric
print("TABLEAU CROISE : COMP ET DV_eletric")

cross_comp_electric = pd.crosstab(
    df["COMP"],
    df["DV_eletric"],
    margins=True
)

display(cross_comp_electric)

# 2. Proportions par ligne
print("\nPROPORTIONS PAR MODALITE DE COMP")

display(
    pd.crosstab(
        df["COMP"],
        df["DV_eletric"],
        normalize="index"
    ).round(4)
)

# 3. Tableau croisé COMP / LPS
print("\nTABLEAU CROISE : COMP ET LPS")

display(
    pd.crosstab(
        df["COMP"],
        df["LPS"],
        margins=True
    )
)

# 4. Statistiques complètes des capteurs selon LPS
print("\nSTATISTIQUES DES CAPTEURS SELON LPS")

display(
    df.groupby("LPS")[sensor_columns]
    .agg(["count", "mean", "median", "std"])
    .round(3)
)
```


### Conclusion de l'analyse des variables binaires


###  Localisation des événements dans le dataset

Les dates des événements sont comparées aux horodatages du dataset.

Pour chaque événement, nous calculons le nombre d'observations disponibles entre sa date de début et sa date de fin.

Cette étape permet de vérifier que les périodes documentées sont bien couvertes par les données.


```python
# Vérification de la présence des observations
event_results = []

for _, event in events.iterrows():
# mask vérifie si chaque valeur dans la colonne timestamps : event["start"] ≤ value_timestamps ≤ event["end"]
    mask = timestamps.between(
        event["start"],
        event["end"],
        inclusive="both"
    )

    event_results.append({
        "event_id": event["event_id"],
        "event_type": event["event_type"],
        "start": event["start"],
        "end": event["end"],
        "duration_hours": event["duration_hours"],
        "observations": int(mask.sum()),
        "first_observation": timestamps[mask].min(),
        "last_observation": timestamps[mask].max()
    })

event_coverage = pd.DataFrame(event_results)

display(event_coverage)
```


###  Contrôle de la continuité des mesures pendant les événements

L'objectif est de vérifier la régularité des acquisitions à l'intérieur des périodes d'incidents documentées.

Pour chaque événement, nous allons calculer :

- Le nombre d'observations.
- L'intervalle moyen entre les mesures.
- Le plus grand intervalle observé.
- Le nombre d'intervalles supérieurs à une minute.

Cette analyse permettra d'identifier les éventuelles interruptions d'acquisition pendant les incidents.


```python
continuity_results = []

for _, event in events.iterrows():

    # Sélection des mesures dans l'événement
    mask = timestamps.between(
        event["start"],
        event["end"],
        inclusive="both"
    )

    event_timestamps = timestamps[mask].sort_values()

    # Intervalles entre les mesures de l'événement
    event_intervals = (
        event_timestamps.diff()
        .dt.total_seconds()
        .dropna()
    )

    continuity_results.append({
        "Event ID": event["event_id"],
        "Observations": len(event_timestamps),
        "Intervalle moyen (s)": event_intervals.mean(),
        "Intervalle médian (s)": event_intervals.median(),
        "Intervalle maximum (s)": event_intervals.max(),
        "Intervalles > 60 s": int(
            (event_intervals > 60).sum()
        )
    })

continuity_report = pd.DataFrame(continuity_results)

display(continuity_report.round(2))
```


###  Localisation des interruptions pendant les événements

Les résultats précédents ont révélé plusieurs interruptions d'acquisition.

Cette étape permet de localiser précisément ces interruptions afin de déterminer si elles sont ponctuelles ou si elles affectent une partie importante de la période documentée.

Pour chaque interruption, nous afficherons :
- L'identifiant de l'événement.
- La date de la dernière mesure avant l'interruption.
- La date de la première mesure après l'interruption.
- La durée de l'interruption en secondes et en minutes.


```python
gap_results = []

for _, event in events.iterrows():

    mask = timestamps.between(
        event["start"],
        event["end"],
        inclusive="both"
    )

    event_timestamps = (
        timestamps[mask]
        .sort_values()
        .reset_index(drop=True)
    )

    intervals = event_timestamps.diff().dt.total_seconds()

    # Repérer les interruptions supérieures à 60 secondes
    gap_indices = intervals[intervals > 60].index

    for idx in gap_indices:

        gap_seconds = intervals.loc[idx]

        gap_results.append({
            "Event ID": event["event_id"],
            "Dernière mesure avant": event_timestamps.loc[idx - 1],
            "Première mesure après": event_timestamps.loc[idx],
            "Interruption (secondes)": gap_seconds,
            "Interruption (minutes)": gap_seconds / 60
        })

gap_report = pd.DataFrame(gap_results)

display(gap_report)
```


###  Conclusion du contrôle de continuité

Le contrôle de continuité des acquisitions a permis d'identifier quatre interruptions supérieures à une minute pendant les événements industriels documentés.

Les observations sont les suivantes :

- Événement 1 : une interruption de 5 min 52 s.
- Événement 2 : aucune interruption supérieure à une minute.
- Événement 3 : deux interruptions de 3 h 47 min et 52 min 21 s.
- Événement 4 : une interruption de 2 min 19 s.

L'événement 3 présente les interruptions les plus importantes.

Ces interruptions doivent être prises en compte dans les analyses temporelles et dans la construction des futures variables cibles.

Une interruption d'acquisition ne constitue pas automatiquement une panne industrielle.

###  Construction de la variable cible target_event

#### Objectif

Créer une variable cible binaire permettant d'identifier les observations situées pendant les quatre événements de fuite d'air documentés.

#### Règles d'annotation

- target_event = 1 : observation comprise dans un intervalle de fuite documenté.
- target_event = 0 : observation située hors des intervalles documentés.

Les bornes temporelles de début et de fin sont incluses.

#### Limites

Cette annotation est construite à partir des rapports d'incidents de l'entreprise.

La valeur 0 ne garantit pas que l'équipement fonctionne normalement. Elle signifie uniquement qu'aucun des quatre événements documentés n'est associé à l'observation.

La variable créée constitue donc une annotation temporelle fondée sur les événements connus.


```python
# Vérifier que les timestamps sont au format datetime
timestamps = pd.to_datetime(
    df["timestamp"],
    errors="raise"
)

# Initialiser la cible à 0
df["target_event"] = 0

# Compter les observations annotées pour chaque événement
annotation_results = []

for _, event in events.iterrows():

    # Sélection des observations dans l'intervalle
    mask = (
        (timestamps >= event["start"]) &
        (timestamps <= event["end"])
    )

    # Affectation de la classe 1
    df.loc[mask, "target_event"] = 1

    annotation_results.append({
        "event_id": event["event_id"],
        "observations_annotated": int(mask.sum())
    })

# Tableau de contrôle par événement
annotation_report = pd.DataFrame(annotation_results)

display(annotation_report)

print(
    "\nNombre total d'observations annotées :",
    int(df["target_event"].sum())
)
```

Après la création de target_event, nous vérifions :

- Le pourcentage de chaque classe.
- La cohérence du nombre d'observations annotées avec les quatre événements.
- Le caractère binaire de la variable.

Cette vérification permet de détecter une éventuelle erreur dans la construction des annotations.


```python
# 1. Répartition des classes
target_distribution = (
    df["target_event"]
    .value_counts()
    .sort_index()
    .rename_axis("target_event")
    .to_frame("observations")
)

target_distribution["percentage"] = (
    target_distribution["observations"]
    / len(df) * 100
)

display(target_distribution.round(3))

# 2. Vérification du nombre total d'observations annotées
print(
    "\nTotal des observations annotées :",
    int(df["target_event"].sum())
)

# 3. Vérification des valeurs autorisées
print(
    "\nValeurs présentes dans target_event :",
    sorted(df["target_event"].unique())
)

# 4. Contrôle de cohérence
assert set(df["target_event"].unique()).issubset({0, 1})

assert int(df["target_event"].sum()) == 29954

assert int(annotation_report["observations_annotated"].sum()) == 29954

print("\nValidation de target_event réussie.")
```

L'objectif est d'identifier les variables présentant des relations statistiques particulièrement fortes.

Une forte corrélation peut signaler une redondance potentielle entre deux variables.

Cette analyse permettra de repérer les variables qui méritent une investigation complémentaire avant la sélection des features.

Aucune variable ne sera supprimée automatiquement sur la seule base de sa corrélation.


```python
# Matrice de corrélation des variables explicatives
feature_correlation = df[feature_columns].corr()

# Extraire uniquement les paires sans doublons
upper_triangle = feature_correlation.where(
    np.triu(
        np.ones(feature_correlation.shape),
        k=1
    ).astype(bool)
)

correlation_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

correlation_pairs.columns = [
    "Variable 1",
    "Variable 2",
    "Corrélation"
]

# Intensité absolue
correlation_pairs["Intensité"] = (
    correlation_pairs["Corrélation"].abs()
)

# Trier les relations les plus fortes
correlation_pairs = correlation_pairs.sort_values(
    by="Intensité",
    ascending=False
)

display(correlation_pairs.head(15).round(4))
```

L'objectif est de distinguer les variables strictement identiques des variables fortement associées, mais qui conservent des différences.


```python
binary_pairs = [
    ("COMP", "MPG"),
    ("COMP", "DV_eletric"),
    ("DV_eletric", "MPG")
]

for col1, col2 in binary_pairs:

    print(f"\n{'=' * 50}")
    print(f"COMPARAISON : {col1} / {col2}")
    print(f"{'=' * 50}")

    # Tableau croisé
    display(pd.crosstab(df[col1], df[col2]))

    # Taux de valeurs identiques
    agreement = (df[col1] == df[col2]).mean() * 100

    # Taux de valeurs opposées
    disagreement = (df[col1] != df[col2]).mean() * 100

    print(f"Taux de concordance : {agreement:.3f} %")
    print(f"Taux de discordance : {disagreement:.3f} %")
```

### Analyse des variables binaires selon target_event

L'objectif est d'étudier la répartition des états binaires des capteurs selon la cible.

Pour chaque variable, nous calculons :

- Le nombre d'observations de chaque état.
- La proportion de chaque état dans la classe 0.
- La proportion de chaque état dans la classe 1.

###  Mesure des associations statistiques

####  Variables binaires et cible

Nous utilisons le V de Cramér pour mesurer l'intensité de l'association entre chaque variable binaire et target_event.

Cette mesure permet de comparer les associations sans dépendre uniquement des effectifs bruts.

Le V de Cramér ne renseigne pas sur le sens de l'association. Les proportions par classe restent nécessaires pour interpréter les résultats.

Les résultats sont considérés comme descriptifs, compte tenu de la dépendance temporelle des observations.


```python
from scipy.stats import chi2_contingency
import numpy as np

cramer_results = []

for column in binary_columns:

    # Tableau de contingence
    contingency_table = pd.crosstab(
        df[column],
        df["target_event"]
    )

    # Calcul du khi-deux sans correction de continuité
    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table,
        correction=False
    )

    n = contingency_table.to_numpy().sum()

    # Calcul du V de Cramér
    min_dimension = min(
        contingency_table.shape[0] - 1,
        contingency_table.shape[1] - 1
    )

    cramer_v = np.sqrt(
        chi2 / (n * min_dimension)
    )

    cramer_results.append({
        "Variable": column,
        "V de Cramer": cramer_v
    })

cramer_df = pd.DataFrame(cramer_results)

cramer_df = cramer_df.sort_values(
    by="V de Cramer",
    ascending=False
)

display(cramer_df.round(4))
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=cramer_df,
    x="V de Cramer",
    y="Variable"
)

plt.title("Association des variables binaires avec target_event")
plt.xlabel("V de Cramér")
plt.ylabel("Variable")
plt.xlim(0, 1)

plt.tight_layout()
plt.savefig(f"Association_viriables_binaires_vs_target_V-Cramer.png")
plt.show()
```


###  Taille d'effet des variables continues

Nous utilisons le d de Cohen pour mesurer l'amplitude de la différence entre les moyennes des classes 1 et 0, en tenant compte de la dispersion des observations.

Une valeur positive indique une moyenne plus élevée pendant les événements documentés.

Une valeur négative indique une moyenne plus faible pendant les événements documentés.

Cette mesure est complétée par la différence des médianes afin de tenir compte des distributions asymétriques et des valeurs extrêmes.

Les seuils d'interprétation sont indicatifs et ne constituent pas des règles universelles.


```python
continuous_results = []

for column in sensor_columns:

    group_0 = df.loc[
        df["target_event"] == 0, column
    ]

    group_1 = df.loc[
        df["target_event"] == 1, column
    ]

    n0 = len(group_0)
    n1 = len(group_1)

    mean_0 = group_0.mean()
    mean_1 = group_1.mean()

    std_0 = group_0.std()
    std_1 = group_1.std()

    # Écart-type combiné
    pooled_std = np.sqrt(
        (
            (n0 - 1) * std_0**2
            + (n1 - 1) * std_1**2
        ) / (n0 + n1 - 2)
    )

    # D de Cohen
    cohens_d = (
        (mean_1 - mean_0) / pooled_std
        if pooled_std != 0 else np.nan
    )

    median_0 = group_0.median()
    median_1 = group_1.median()

    continuous_results.append({
        "Variable": column,
        "Moyenne classe 0": mean_0,
        "Moyenne classe 1": mean_1,
        "Ecart des médianes": median_1 - median_0,
        "D de Cohen": cohens_d,
        "|D de Cohen|": abs(cohens_d)
    })

cohens_df = pd.DataFrame(continuous_results)

cohens_df = cohens_df.sort_values(
    by="|D de Cohen|",
    ascending=False
)

display(cohens_df.round(4))
```


##  Protocole expérimental : Leave-One-Event-Out

###  Définition de la stratégie

Nous adoptons une validation Leave-One-Event-Out (LOEO).

Le dataset contient quatre événements de fuite documentés.

Le protocole consiste à réserver successivement un événement complet pour le test, tandis que les trois autres événements sont utilisés pour l'apprentissage.

Cette méthode permet d'étudier la capacité du modèle à détecter un événement qui n'a pas été observé pendant son entraînement.

Les observations négatives devront également être réparties entre apprentissage et test afin de permettre une évaluation complète des performances.


```python
validation_events = pd.DataFrame({
    "event_id": [1, 2, 3, 4],

    "start": [
        "2020-04-18 00:00:00",
        "2020-05-29 23:30:00",
        "2020-06-05 10:00:00",
        "2020-07-15 14:30:00"
    ],

    "end": [
        "2020-04-18 23:59:00",
        "2020-05-30 06:00:00",
        "2020-06-07 14:30:00",
        "2020-07-15 19:00:00"
    ]
})

validation_events["start"] = pd.to_datetime(
    validation_events["start"]
)

validation_events["end"] = pd.to_datetime(
    validation_events["end"]
)

validation_events["duration_hours"] = (
    validation_events["end"] -
    validation_events["start"]
).dt.total_seconds() / 3600

display(validation_events)
```

```python
assert len(validation_events) == 4

assert validation_events["event_id"].is_unique

assert (
    validation_events["start"] <
    validation_events["end"]
).all()

events_sorted = validation_events.sort_values("start")

overlaps = (
    events_sorted["start"].iloc[1:].reset_index(drop=True)
    <=
    events_sorted["end"].iloc[:-1].reset_index(drop=True)
)

assert not overlaps.any()

# Vérifier la présence d'observations positives
for _, event in validation_events.iterrows():

    event_mask = (
        (df["timestamp"] >= event["start"]) &
        (df["timestamp"] <= event["end"])
    )

    positive_count = (
        df.loc[event_mask, "target_event"] == 1
    ).sum()

    print(
        f"Événement {event['event_id']} : "
        f"{positive_count} observations positives"
    )

print("\nTous les contrôles du registre sont validés.")
```


### Construction des fenêtres de test

Chaque événement documenté est associé à une fenêtre de test comprenant :

- Les 24 heures précédant l'événement.
- La totalité de l'événement.
- Les 24 heures suivant l'événement.

Toutes les observations présentes dans cette fenêtre sont affectées au jeu de test.

Les observations situées en dehors de cette fenêtre sont affectées au jeu d'entraînement.

Cette stratégie permet d'évaluer le modèle sur un événement entièrement absent de son apprentissage, avec un contexte temporel négatif avant et après l'événement.


```python
TEST_CONTEXT = pd.Timedelta(hours=24)

validation_events["test_start"] = (
    validation_events["start"] - TEST_CONTEXT
)

validation_events["test_end"] = (
    validation_events["end"] + TEST_CONTEXT
)

display(
    validation_events[
        [
            "event_id",
            "test_start",
            "start",
            "end",
            "test_end"
        ]
    ]
)
```


###  Contrôle du chevauchement des fenêtres

Les fenêtres de test ne doivent pas se chevaucher.

Un chevauchement pourrait entraîner l'inclusion d'un événement dans le test d'un autre événement, ce qui compromettrait l'indépendance des expériences.

Nous vérifions donc que chaque fenêtre se termine strictement avant le début de la suivante.


```python
events_windows_sorted = validation_events.sort_values(
    "test_start"
).reset_index(drop=True)

window_overlaps = []

for i in range(len(events_windows_sorted) - 1):

    current_end = events_windows_sorted.loc[i, "test_end"]
    next_start = events_windows_sorted.loc[i + 1, "test_start"]

    overlap = next_start <= current_end

    window_overlaps.append(overlap)

    print(
        f"Événements "
        f"{events_windows_sorted.loc[i, 'event_id']} et "
        f"{events_windows_sorted.loc[i + 1, 'event_id']} : "
        f"{'CHEVAUCHEMENT' if overlap else 'Aucun chevauchement'}"
    )

assert not any(window_overlaps), (
    "Erreur : les fenêtres de test se chevauchent."
)

print("\nContrôle des fenêtres réussi.")
```


###  Attribution des observations aux événements

Chaque observation positive est associée à l'identifiant de l'événement correspondant à son timestamp.

Cette colonne event_id est une information de traçabilité destinée à construire les folds.

Elle ne constitue pas une variable explicative et ne sera pas transmise au modèle.


```python
# Vérifier le format des timestamps
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Créer une colonne de traçabilité
df["event_id"] = pd.Series(
    pd.NA,
    index=df.index,
    dtype="Int64"
)

# Attribuer chaque observation à son événement
for _, event in validation_events.iterrows():

    event_mask = (
        (df["timestamp"] >= event["start"]) &
        (df["timestamp"] <= event["end"])
    )

    df.loc[event_mask, "event_id"] = int(event["event_id"])

# Calculer les effectifs par événement
event_counts = (
    df[df["target_event"] == 1]
    .groupby("event_id")
    .size()
    .rename("observations")
    .reset_index()
)

display(event_counts)

# Contrôle 1 : toutes les observations positives sont identifiées
assert df.loc[
    df["target_event"] == 1,
    "event_id"
].notna().all()

# Contrôle 2 : aucune observation négative n'est identifiée
assert df.loc[
    df["target_event"] == 0,
    "event_id"
].isna().all()

# Contrôle 3 : les effectifs correspondent au registre
for _, event in validation_events.iterrows():

    event_id = int(event["event_id"])

    expected_count = int(
        (
            (df["timestamp"] >= event["start"]) &
            (df["timestamp"] <= event["end"])
        ).sum()
    )

    observed_count = int(
        event_counts.loc[
            event_counts["event_id"] == event_id,
            "observations"
        ].iloc[0]
    )

    assert observed_count == expected_count

print("\nAttribution des événements validée.")
```


### Construction des quatre folds LOEO

Nous construisons quatre expériences indépendantes.

Pour chaque fold, un événement est entièrement réservé au test, avec ses 24 heures de contexte avant et après.

Les autres événements documentés sont conservés dans l'entraînement.

Les observations négatives sont réparties selon leur appartenance temporelle à la fenêtre de test.

Les index d'origine sont conservés pour assurer la traçabilité des observations.


```python
folds = {}

for _, event in validation_events.iterrows():

    event_id = int(event["event_id"])

    test_start = event["test_start"]
    test_end = event["test_end"]

    # Toutes les observations de la fenêtre de test
    test_mask = (
        (df["timestamp"] >= test_start) &
        (df["timestamp"] <= test_end)
    )

    # Les observations situées hors de la fenêtre
    train_mask = ~test_mask

    X_train = df.loc[
        train_mask,
        feature_columns
    ].copy()

    y_train = df.loc[
        train_mask,
        target_column
    ].copy()

    X_test = df.loc[
        test_mask,
        feature_columns
    ].copy()

    y_test = df.loc[
        test_mask,
        target_column
    ].copy()

    # Conservation des identifiants pour les contrôles
    train_event_ids = df.loc[train_mask, "event_id"].copy()
    test_event_ids = df.loc[test_mask, "event_id"].copy()

    folds[event_id] = {
        "X_train": X_train,
        "y_train": y_train,
        "X_test": X_test,
        "y_test": y_test,
        "train_event_ids": train_event_ids,
        "test_event_ids": test_event_ids,
        "test_start": test_start,
        "test_end": test_end
    }

print(f"Nombre de folds construits : {len(folds)}")
```


###  Contrôle de la séparation des folds

Nous vérifions automatiquement l'intégrité de chaque expérience.

Un fold est valide si :

- L'événement réservé est entièrement présent dans le test.
- Aucun autre événement positif n'est présent dans ce test.
- Les autres événements positifs restent dans l'entraînement.
- Les deux classes sont présentes dans le test.
- Les ensembles d'entraînement et de test ne partagent aucune observation.
- Les variables explicatives sont identiques dans les deux ensembles.

Ces contrôles sont indispensables avant toute phase d'apprentissage.


```python
fold_summary = []

for event_id, fold in folds.items():

    X_train = fold["X_train"]
    y_train = fold["y_train"]

    X_test = fold["X_test"]
    y_test = fold["y_test"]

    train_event_ids = fold["train_event_ids"]
    test_event_ids = fold["test_event_ids"]

    # Effectifs des classes
    train_counts = y_train.value_counts()
    test_counts = y_test.value_counts()

    # Vérification des événements positifs du test
    test_positive_ids = set(
        test_event_ids.dropna().astype(int).unique()
    )

    # Vérification des événements positifs de l'entraînement
    train_positive_ids = set(
        train_event_ids.dropna().astype(int).unique()
    )

    assert test_positive_ids == {event_id}

    assert event_id not in train_positive_ids

    assert len(test_counts) == 2

    assert len(train_counts) == 2

    # Vérification de l'absence de chevauchement des index
    assert X_train.index.intersection(X_test.index).empty

    # Vérification des variables explicatives
    assert list(X_train.columns) == feature_columns
    assert list(X_test.columns) == feature_columns

    # Vérification de l'effectif positif attendu
    expected_positive = int(
        (
            (df["event_id"] == event_id) &
            (df["target_event"] == 1)
        ).sum()
    )

    assert int(test_counts.loc[1]) == expected_positive

    fold_summary.append({
        "Fold": event_id,
        "Événement testé": event_id,
        "Train total": len(X_train),
        "Train positifs": int(train_counts.loc[1]),
        "Train négatifs": int(train_counts.loc[0]),
        "Test total": len(X_test),
        "Test positifs": int(test_counts.loc[1]),
        "Test négatifs": int(test_counts.loc[0]),
        "Index communs": len(
            X_train.index.intersection(X_test.index)
        )
    })

fold_summary_df = pd.DataFrame(fold_summary)

display(fold_summary_df)

print("\nTous les contrôles des quatre folds sont validés.")
```

```python
# Vérification de l'absence de chevauchement entre les tests
for event_id, fold in folds.items():

    test_start = fold["test_start"]
    test_end = fold["test_end"]

    for other_id, other_fold in folds.items():

        if event_id == other_id:
            continue

        other_start = other_fold["test_start"]
        other_end = other_fold["test_end"]

        assert (
            test_end < other_start or
            other_end < test_start
        ), (
            f"Chevauchement entre les folds "
            f"{event_id} et {other_id}"
        )

# Vérification du nombre de folds
assert len(folds) == 4

# Vérification du nombre total d'observations positives
assert (
    fold_summary_df["Test positifs"].sum()
    == df["target_event"].sum()
)

display(fold_summary_df)

print("Nombre de folds :", len(folds))
print("Observations positives totales :", int(df["target_event"].sum()))
print("Contrôle final réussi.")
```

Chaque fold réserve un événement complet au test, accompagné d'une fenêtre de contexte de 24 heures avant et après.

Les contrôles portent sur :

- L'absence d'un même événement positif dans les deux ensembles.
- La présence des deux classes dans chaque jeu de test.
- La conservation des autres événements dans l'entraînement.
- L'absence de partage d'index entre entraînement et test.
- La conservation des 15 variables explicatives.

Cette stratégie permet d'étudier la généralisation à des événements documentés non vus pendant l'apprentissage.

### Analyse de l'équilibre des classes

L'objectif est d'étudier la répartition des classes positives et négatives dans chaque fold.

Cette analyse permet de mesurer le déséquilibre des classes et de vérifier si les différents événements présentent des conditions d'évaluation comparables.

Les proportions sont calculées séparément pour les ensembles d'entraînement et de test.


### Protocole d'évaluation

Les performances de chaque modèle seront évaluées individuellement sur les quatre folds LOEO.

Le Recall constitue la métrique principale, car notre objectif prioritaire est de détecter les événements de fuite.

Les autres métriques permettront d'évaluer la fiabilité des alertes et la capacité globale de discrimination.

Les métriques seront calculées à partir :
- Des prédictions binaires pour Recall, Precision, F1-score et Balanced Accuracy.
- Des probabilités prédites pour Average Precision et ROC-AUC.
- Des prédictions binaires et des valeurs réelles pour la matrice de confusion.

Le seuil de classification sera défini séparément. Il ne sera pas ajusté sur les données de test.


```python
from sklearn.metrics import (
    recall_score,
    precision_score,
    f1_score,
    average_precision_score,
    balanced_accuracy_score,
    roc_auc_score,
    confusion_matrix
)

print("Métriques d'évaluation importées avec succès.")
```

```python
def evaluate_classification(y_true, y_pred, y_proba):

    results = {
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "F1-score": f1_score(y_true, y_pred, zero_division=0),
        "PR-AUC (AP)": average_precision_score(y_true, y_proba),
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_proba)
    }

    matrix = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    return results, matrix
```

```python
#  TEST TECHNIQUE DE LA FONCTION

y_true_demo = [0, 0, 1, 1, 1, 0]

y_proba_demo = [0.10, 0.30, 0.80, 0.65, 0.20, 0.40]

# Seuil provisoire de 0.5
y_pred_demo = [
    int(probability >= 0.5)
    for probability in y_proba_demo
]

demo_results, demo_matrix = evaluate_classification(
    y_true_demo,
    y_pred_demo,
    y_proba_demo
)

display(pd.DataFrame(
    demo_results.items(),
    columns=["Métrique", "Résultat"]
))

print("Matrice de confusion :")
display(pd.DataFrame(
    demo_matrix,
    index=["Réel 0", "Réel 1"],
    columns=["Prédit 0", "Prédit 1"]
))
```


## Sélection du seuil par validation interne

La sélection du seuil sera réalisée exclusivement à partir des données d'entraînement de chaque fold externe.

Nous utilisons une validation interne de type Leave-One-Event-Out.

Pour chaque événement externe, les trois événements restants sont successivement utilisés comme événements de validation.

Chaque validation interne utilise une fenêtre de 24 heures avant et après l'événement considéré.

Les observations hors de cette fenêtre constituent l'ensemble d'entraînement interne.

Les prédictions obtenues sur les validations internes serviront ensuite à sélectionner le seuil de classification.

Le jeu de test externe ne sera jamais utilisé pour cette sélection.


```python
inner_folds = {}

for outer_event_id, outer_fold in folds.items():

    # Observations disponibles dans le train externe
    outer_train_index = outer_fold["X_train"].index

    outer_train_data = df.loc[outer_train_index]

    # Les événements disponibles pour la validation interne
    remaining_events = validation_events[
        validation_events["event_id"] != outer_event_id
    ]

    inner_folds[outer_event_id] = {}

    for _, inner_event in remaining_events.iterrows():

        inner_event_id = int(inner_event["event_id"])

        inner_start = inner_event["test_start"]
        inner_end = inner_event["test_end"]

        # Validation interne : fenêtre temporelle de l'événement
        inner_val_mask = (
            (outer_train_data["timestamp"] >= inner_start) &
            (outer_train_data["timestamp"] <= inner_end)
        )

        # Entraînement interne : observations hors de la fenêtre
        inner_train_mask = ~inner_val_mask

        inner_train_index = outer_train_data.index[inner_train_mask]
        inner_val_index = outer_train_data.index[inner_val_mask]

        inner_folds[outer_event_id][inner_event_id] = {
            "train_index": inner_train_index,
            "val_index": inner_val_index
        }

print("Validations internes construites.")
```

```python

inner_summary = []

for outer_event_id, inner_splits in inner_folds.items():

    for inner_event_id, split in inner_splits.items():

        train_index = split["train_index"]
        val_index = split["val_index"]

        y_inner_train = df.loc[train_index, "target_event"]
        y_inner_val = df.loc[val_index, "target_event"]

        val_event_ids = set(
            df.loc[val_index, "event_id"]
            .dropna()
            .astype(int)
            .unique()
        )

        # Absence de chevauchement des index
        assert train_index.intersection(val_index).empty

        # L'événement de validation doit être présent
        assert val_event_ids == {inner_event_id}

        # L'événement de validation ne doit pas être dans le train
        train_event_ids = set(
            df.loc[train_index, "event_id"]
            .dropna()
            .astype(int)
            .unique()
        )

        assert inner_event_id not in train_event_ids

        # Les deux classes doivent être présentes en validation
        assert y_inner_val.nunique() == 2

        # Les deux classes doivent être présentes en entraînement
        assert y_inner_train.nunique() == 2

        inner_summary.append({
            "Fold externe": outer_event_id,
            "Événement validation": inner_event_id,
            "Train interne": len(train_index),
            "Validation interne": len(val_index),
            "Positifs validation": int(y_inner_val.sum()),
            "Négatifs validation": int((y_inner_val == 0).sum()),
            "Index communs": len(
                train_index.intersection(val_index)
            )
        })

inner_summary_df = pd.DataFrame(inner_summary)

display(inner_summary_df)

print("Nombre de validations internes :", len(inner_summary_df))
print("Tous les contrôles internes sont validés.")
```


### Préparation du modèle Baseline

Nous choisissons la régression logistique comme premier modèle de référence.

Il s'agit d'un modèle de classification binaire permettant d'estimer la probabilité d'appartenance à la classe positive.

Les 15 variables explicatives candidates sont conservées.

Les colonnes de traçabilité et d'identification des événements sont exclues afin d'éviter leur utilisation accidentelle par le modèle.

Le prétraitement et l'apprentissage seront intégrés dans un pipeline Scikit-learn.


```python
feature_columns = [
    "TP2",
    "TP3",
    "H1",
    "DV_pressure",
    "Reservoirs",
    "Oil_temperature",
    "Motor_current",
    "COMP",
    "DV_eletric",
    "Towers",
    "MPG",
    "LPS",
    "Pressure_switch",
    "Oil_level",
    "Caudal_impulses"
]

target_column = "target_event"

excluded_columns = [
    "timestamp",
    "Unnamed: 0",
    "event_id"
]

print("Nombre de variables explicatives :", len(feature_columns))
print("Variable cible :", target_column)
print("Variables exclues :", excluded_columns)

assert len(feature_columns) == 15
assert target_column in df.columns

assert all(
    column in df.columns
    for column in feature_columns
)

assert not set(feature_columns).intersection(
    excluded_columns + [target_column]
)

print("\nDéfinition des variables validée.")
```


###  Contrôle de qualité des variables

Avant l'apprentissage, nous vérifions la nature numérique des variables, l'absence de valeurs manquantes et infinies, ainsi que la validité de la variable cible.

Ces contrôles permettent de détecter les problèmes de données avant la construction du pipeline.


```python
X_check = df[feature_columns]
y_check = df[target_column]

# Vérification des types numériques
assert all(
    pd.api.types.is_numeric_dtype(X_check[col])
    for col in feature_columns
)

# Vérification des valeurs manquantes
missing_values = X_check.isna().sum()

# Vérification des valeurs infinies
infinite_values = np.isinf(X_check).sum()

# Vérification de la cible
assert y_check.notna().all()
assert set(y_check.unique()) == {0, 1}

# Synthèse
quality_report = pd.DataFrame({
    "Type": X_check.dtypes.astype(str),
    "Valeurs manquantes": missing_values,
    "Valeurs infinies": infinite_values,
    "Valeurs distinctes": X_check.nunique()
})

display(quality_report)

assert missing_values.sum() == 0
assert infinite_values.sum() == 0

print("\nContrôle qualité des variables validé.")
```


###  Construction du pipeline

Le modèle Baseline utilise :

- StandardScaler pour normaliser les variables.
- LogisticRegression pour réaliser la classification.
- class_weight="balanced" pour tenir compte du déséquilibre des classes.

Le pipeline garantit que la normalisation est ajustée uniquement sur les données utilisées pour entraîner le modèle.


```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

def build_baseline_pipeline():

    pipeline = Pipeline([
        ("scaler", StandardScaler()),

        ("classifier", LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            solver="lbfgs",
            random_state=42
        ))
    ])

    return pipeline


baseline_pipeline = build_baseline_pipeline()

print(baseline_pipeline)
```

```python
assert isinstance(baseline_pipeline, Pipeline)

assert isinstance(
    baseline_pipeline.named_steps["scaler"],
    StandardScaler
)

assert isinstance(
    baseline_pipeline.named_steps["classifier"],
    LogisticRegression
)

classifier = baseline_pipeline.named_steps["classifier"]

assert classifier.class_weight == "balanced"
assert classifier.max_iter == 2000
assert classifier.solver == "lbfgs"

print("Pipeline Baseline correctement configuré.")
print("Variables explicatives :", len(feature_columns))
print("Nombre de classes :", y_check.nunique())
print("Entraînement non effectué.")
```


###  Entraînement et prédictions internes

Pour chaque fold externe, nous entraînons trois modèles internes.

Chaque modèle est entraîné sur les observations du train interne et prédit les probabilités de la validation interne correspondante.

Les prédictions sont réalisées avec predict_proba afin de conserver les probabilités de la classe positive.

Les résultats sont regroupés pour permettre une sélection du seuil de classification sans utiliser le test externe.


```python
oof_predictions = {}

for outer_event_id, inner_splits in inner_folds.items():

    fold_predictions = []

    for inner_event_id, split in inner_splits.items():

        train_index = split["train_index"]
        val_index = split["val_index"]

        # Préparation des données internes
        X_train = df.loc[train_index, feature_columns]
        y_train = df.loc[train_index, target_column]

        X_val = df.loc[val_index, feature_columns]
        y_val = df.loc[val_index, target_column]

        # Nouveau pipeline pour chaque validation
        model = build_baseline_pipeline()

        # Entraînement interne
        model.fit(X_train, y_train)

        # Probabilités de la classe positive
        probabilities = model.predict_proba(X_val)[:, 1]

        # Conservation des résultats
        predictions = pd.DataFrame({
            "outer_fold": outer_event_id,
            "validation_event": inner_event_id,
            "y_true": y_val.to_numpy(),
            "y_probability": probabilities
        }, index=val_index)

        fold_predictions.append(predictions)

    # Regroupement des trois validations internes
    oof_predictions[outer_event_id] = pd.concat(
        fold_predictions
    )

print("Entraînements internes terminés.")

for fold_id, predictions in oof_predictions.items():

    print(
        f"Fold {fold_id} : "
        f"{len(predictions)} prédictions internes"
    )
```

```python
prediction_checks = []

for fold_id, predictions in oof_predictions.items():

    assert predictions["y_probability"].notna().all()

    assert predictions["y_probability"].between(0, 1).all()

    assert set(predictions["y_true"].unique()) == {0, 1}

    assert predictions.index.is_unique

    prediction_checks.append({
        "Fold externe": fold_id,
        "Observations prédites": len(predictions),
        "Positifs": int(predictions["y_true"].sum()),
        "Négatifs": int((predictions["y_true"] == 0).sum()),
        "Probabilité minimale": predictions["y_probability"].min(),
        "Probabilité maximale": predictions["y_probability"].max(),
        "Index dupliqués": predictions.index.duplicated().sum()
    })

prediction_checks_df = pd.DataFrame(prediction_checks)

display(prediction_checks_df)

print("Nombre de folds :", len(oof_predictions))
print("Contrôle des prédictions internes terminé.")
```


###  Optimisation du seuil de classification

Pour chaque fold externe, nous recherchons le seuil maximisant le F2-score à partir des prédictions internes hors échantillon.

Une grille de seuils comprise entre 0.01 et 0.99 est évaluée.

En cas d'égalité du F2-score, nous retenons le seuil le plus élevé afin de privilégier la précision et de limiter les fausses alertes.

Le seuil sélectionné est ensuite conservé pour l'évaluation externe correspondante.

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score,
    confusion_matrix
)

thresholds = np.arange(0.01, 1.00, 0.01)

threshold_results = {}
best_thresholds = {}

for fold_id, predictions in oof_predictions.items():

    y_true = predictions["y_true"].to_numpy()
    y_prob = predictions["y_probability"].to_numpy()

    fold_results = []

    for threshold in thresholds:

        y_pred = (y_prob >= threshold).astype(int)

        precision = precision_score(
            y_true, y_pred, zero_division=0
        )

        recall = recall_score(
            y_true, y_pred, zero_division=0
        )

        f2 = fbeta_score(
            y_true, y_pred, beta=2, zero_division=0
        )

        tn, fp, fn, tp = confusion_matrix(
            y_true, y_pred, labels=[0, 1]
        ).ravel()

        fold_results.append({
            "threshold": threshold,
            "precision": precision,
            "recall": recall,
            "f2_score": f2,
            "false_positives": fp,
            "false_negatives": fn,
            "true_positives": tp,
            "true_negatives": tn
        })

    results_df = pd.DataFrame(fold_results)

    # En cas d'égalité, retenir le seuil le plus élevé
    best_row = (
        results_df
        .sort_values(
            ["f2_score", "threshold"],
            ascending=[False, False]
        )
        .iloc[0]
    )

    threshold_results[fold_id] = results_df
    best_thresholds[fold_id] = float(best_row["threshold"])

print("Sélection des seuils terminée.")
```

```python
threshold_summary = []

for fold_id, results_df in threshold_results.items():

    best_threshold = best_thresholds[fold_id]

    best_row = results_df[
        results_df["threshold"] == best_threshold
    ].iloc[0]

    threshold_summary.append({
        "Fold externe": fold_id,
        "Seuil sélectionné": best_threshold,
        "Precision": best_row["precision"],
        "Recall": best_row["recall"],
        "F2-score": best_row["f2_score"],
        "Fausses alertes": int(best_row["false_positives"]),
        "Fuites manquées": int(best_row["false_negatives"])
    })

threshold_summary_df = pd.DataFrame(threshold_summary)

display(
    threshold_summary_df.round(4)
)
```

```python
fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 9)
)

axes = axes.flatten()

for i, (fold_id, results_df) in enumerate(
    threshold_results.items()
):

    ax = axes[i]

    ax.plot(
        results_df["threshold"],
        results_df["f2_score"],
        label="F2-score"
    )

    ax.axvline(
        best_thresholds[fold_id],
        linestyle="--",
        label=f"Seuil retenu : {best_thresholds[fold_id]:.2f}"
    )

    ax.set_title(f"Fold externe {fold_id}")
    ax.set_xlabel("Seuil de classification")
    ax.set_ylabel("F2-score")
    ax.grid(True, alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.savefig(f"Evaluation_F2-Score.png")
plt.show()
```


###  Entraînement final et évaluation externe

Pour chaque fold externe, le pipeline de régression logistique est réentraîné sur l'intégralité du jeu d'entraînement externe.

Le seuil de classification est celui sélectionné lors de la validation interne.

Le modèle prédit ensuite les probabilités sur le jeu de test externe.

Les performances sont évaluées à l'aide des métriques suivantes :

- Precision
- Recall
- F1-score
- F2-score
- PR-AUC (Average Precision)
- ROC-AUC
- Balanced Accuracy

Les matrices de confusion sont également conservées.

Aucun ajustement du modèle ou du seuil ne sera effectué à partir des résultats du test externe.


```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    average_precision_score,
    roc_auc_score,
    balanced_accuracy_score,
    confusion_matrix
)

outer_results = {}
outer_models = {}
outer_confusion_matrices = {}

for fold_id, outer_fold in folds.items():

    # Récupération des index externes
    train_index = outer_fold["X_train"].index
    test_index = outer_fold["X_test"].index

    # Contrôle de séparation
    assert train_index.intersection(test_index).empty

    # Préparation des données
    X_train = df.loc[train_index, feature_columns]
    y_train = df.loc[train_index, target_column]

    X_test = df.loc[test_index, feature_columns]
    y_test = df.loc[test_index, target_column]

    # Récupération du seuil sélectionné en interne
    threshold = best_thresholds[fold_id]

    # Construction et entraînement du modèle final
    model = build_baseline_pipeline()
    model.fit(X_train, y_train)

    # Probabilités de la classe positive
    y_probability = model.predict_proba(X_test)[:, 1]

    # Classification avec le seuil interne
    y_pred = (y_probability >= threshold).astype(int)

    # Calcul des métriques
    metrics = {
        "Fold externe": fold_id,
        "Seuil": threshold,
        "Observations test": len(y_test),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1-score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "F2-score": fbeta_score(
            y_test, y_pred, beta=2, zero_division=0
        ),
        "PR-AUC (AP)": average_precision_score(
            y_test, y_probability
        ),
        "ROC-AUC": roc_auc_score(
            y_test, y_probability
        ),
        "Balanced Accuracy": balanced_accuracy_score(
            y_test, y_pred
        )
    }

    # Matrice de confusion
    cm = confusion_matrix(
        y_test, y_pred, labels=[0, 1]
    )

    outer_results[fold_id] = metrics
    outer_models[fold_id] = model
    outer_confusion_matrices[fold_id] = cm

# Tableau de synthèse
outer_results_df = pd.DataFrame(
    outer_results.values()
)

display(outer_results_df.round(4))

print("Évaluation des quatre folds externes terminée.")
```
